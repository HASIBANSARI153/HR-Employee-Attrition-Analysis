*** Data cleaning using pandas ***

```python
df_clean=df.copy()
```

```python
df_clean=df_clean.drop_duplicates()
```

```python
df_clean.isnull().sum().sum()
```

```python
df_clean=df_clean.drop(columns=['EmployeeCount','Over18','StandardHours'])
```

```python
df_clean.shape
```

* Creating new analytical columns

```python
# creating age group

df_clean['Age_Group']=pd.cut(df_clean['Age'],bins=[17, 25, 35, 45, 55, 100],labels=['18-25','26-35','36-45','46-55','56+'])
```

```python
# create attrition flag

df_clean['Attrition_Flag']=(df_clean['Attrition'].map({'Yes':1, 'No':0 }))
```

```python
# creating experience group

df_clean['Experience_Group'] = pd.cut(
    df_clean['YearsAtCompany'],
    bins=[-1, 4, 10, 40],
    labels=['Less than 5 Years',
            '5-10 Years',
            'Above 10 Years']
)

```

```python
df_clean[
    ['EmployeeName', 'Age_Group',
     'Attrition_Flag', 'Experience_Group']
].head(10)
```

*Perform HR Data Analysis Using Pandas*

```python
# total employees

df_clean['EmployeeNumber'].nunique()
```

```python
# total employees who left

df_clean[
    df_clean['Attrition'] == 'Yes'
].shape[0]
```

```python
# Attrition percentage

attrition_rate =(
    df_clean['Attrition_Flag'].mean()*100)
print(round(attrition_rate,2))
```

```python
# department - wise attrition

df_clean.groupby('Department')['Attrition_Flag'].mean().mul(100).round(2)
```

```python
# Gender -Wise Attrition

df_clean.groupby('Gender')['Attrition_Flag'].mean().mul(100).round(2)
```

```python
# Average Salary by department

df_clean.groupby('Department')['MonthlyIncome'].mean().round(2)
```

```python
# Atterition by overtime

df_clean.groupby('OverTime')['Attrition_Flag'].mean().mul(100).round(2)
```

```python
# Attrition by marital status

df_clean.groupby('MaritalStatus')['Attrition_Flag'].mean().mul(100).round(2)
```

```python
# Attrition by job satisfaction

df_clean.groupby('JobSatisfaction')['Attrition_Flag'].mean().mul(100).round(2)
```

```python
# Average monthely income by gender

df_clean.groupby('Gender')['MonthlyIncome'].mean().round(2)
```

*** Creating charts ***

```python
# attrition distribution

df_clean['Attrition'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Employee Attrition Distribution')
plt.ylabel('')
plt.show()
```

```python
# department-wise attrition

sns.countplot(
    data=df_clean,
    x='Department',
    hue='Attrition'
)

plt.title('Department-wise Employee Attrition')
plt.xticks(rotation=15)
plt.show()
```

```python
# age group analysis

sns.countplot(
    data=df_clean,
    x='Age_Group',
    hue='Attrition'
)

plt.title('Attrition by Age Group')
plt.show()
```

```python
# overtime analysis

sns.countplot(
    data=df_clean,
    x='OverTime',
    hue='Attrition'
)

plt.title('Overtime vs Employee Attrition')
plt.show()
```

```python
# salary distribution

sns.boxplot(
    data=df_clean,
    x='Attrition',
    y='MonthlyIncome'
)

plt.title('Salary Distribution by Attrition')
plt.show()
```
